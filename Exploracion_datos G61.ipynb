```python
#Importar archivo
df_ventasNum = pd.read_excel("ventasNum2025.xlsx")
df_ventasNum.head()
```

```python
# Renombrar columnas
df_ventasNum = df_ventasNum.rename(columns={
    'Ingresos por productos (MXN)': 'Ingreso_X_producto', 
    'Cargo por venta e impuestos': 'Cargos_impuestos',
    'Costos de envío': 'Costo_envio',
    'Total (MXN)': 'Total',
    'Venta por publicidad': 'Venta_publicidad',
    'Canal de venta': 'Canal_venta',
    'Canal de venta': 'Canal_venta',
    'Precio unitario de venta de la publicación (MXN)': 'Precio_unitario',
    'Municipio/Alcaldía': 'Municipio_Alcaldia',
    'Código postal':'CP',
    'Forma de entrega':'Entrega',
    'Fecha Venta':'FechaVenta'
})

df_ventasNum.head()
```

```python
#Verifica tipo de datos
df_ventasNum.info()
```

```python
#Conocer distribución de variables numéricas
df_ventasNum.describe()
```

```python
#Modificación de valores númericos de cargo por venta e impuestos y costros de envío
#ya que se encuentran en valores negativos

#Valores positivos de "Cargo por venta e impuestos"
df_ventasNum['Cargos_impuestos'] = df_ventasNum['Cargos_impuestos'].abs()
#Valores positivos de "Costo de envío"
df_ventasNum['Costo_envio'] = df_ventasNum['Costo_envio'].abs()

df_ventasNum.head()
```

```python
# Seleccionar solo las columnas de tipo objeto o categóricas
columnas_categoricas = df_ventasNum.select_dtypes(include=['object', 'category']).columns

# Iterar e imprimir los valores únicos de cada una
for col in columnas_categoricas:
    print(f"Columna: {col}")
    print(df_ventasNum[col].unique())
    print("-" * 40)
```

```python
#Guardar nuevo excel
df_ventasNum.to_excel('ventasNum_modified.xlsx', sheet_name='Datos_Principales', index=False)
```
